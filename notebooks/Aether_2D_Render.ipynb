```python
# Cell 1 — Aether Studio v2 Preflight & Bootstrapper (Colab CPU or T4 GPU)
import pathlib, subprocess, sys, urllib.request
BASE = "https://raw.githubusercontent.com/aether-autonomus-cell/aether-studio-app/main"
pathlib.Path("/content/colab_worker.py").write_bytes(
    urllib.request.urlopen(f"{BASE}/colab_worker.py", timeout=30).read()
)
subprocess.run(
    [sys.executable, "/content/colab_worker.py", "preflight", "--base-url", BASE],
    check=True,
)
```

```python
# Cell 2 — Render Hindi + English MP4s & Auto-Save to Google Drive
import pathlib, subprocess, sys
MODE = "mode-a"  # Change to "mode-b" if pairing with Cloud Queue
if MODE == "mode-a":
    JOB_JSON = r'''PASTE_AETHER_JOB_JSON_HERE'''
    job_path = pathlib.Path("/content/aether_job.json")
    if "PASTE_AETHER_JOB_JSON_HERE" not in JOB_JSON:
        job_path.write_text(JOB_JSON, encoding="utf-8")
    subprocess.run([sys.executable, "/content/colab_worker.py", "render", "--job", str(job_path)], check=True)
else:
    import getpass
    origin = input("Vercel Studio Origin (https://...): ").strip()
    code = getpass.getpass("12-char Pairing Code: ").strip()
    subprocess.run([sys.executable, "/content/colab_worker.py", "render", "--origin", origin, "--pair-code", code], check=True)
```

```python
# Cell 3 — Optional Dual-Channel Direct YouTube Publish
import pathlib, subprocess, sys
MANIFEST_JSON = r'''PASTE_PUBLISH_MANIFEST_JSON_HERE'''
manifest_path = pathlib.Path("/content/publish_manifest.json")
if "PASTE_PUBLISH_MANIFEST_JSON_HERE" not in MANIFEST_JSON:
    manifest_path.write_text(MANIFEST_JSON, encoding="utf-8")
subprocess.run([sys.executable, "/content/colab_worker.py", "publish", "--manifest", str(manifest_path)], check=True)
```
